```python
# latihan 4
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir memakai indeks
print(f"pembacaan pertama: {tegangan_harian[0]}")
print(f"pembacaan terakhir: {tegangan_harian[-1]}")

# (b) Tambahkan satu pembacaan baru 222.0 dangan append()
tegangan_harian.append(222.0)
print(f"Daftar tegangan_harian setelah ditambahkan: {tegangan_harian}")

# (c) Cetak jumlah total data dengan len()
print(f"jumlah total data: {len(tegangan_harian)}")
```

```python
# latihan 5

# buat dictionary spesifikasi motor listrik

motor_spec = {"nama": "motor induksi AC",
              "daya_kw": 15.0,
              "tegangan_V": 380,
              "rpm": 1450}

# Cetak salah satu nilai lewat key_nya
print(f"Nama Motor: {motor_spec['nama']}")

# Ubah nilai rpm
motor_spec['rpm'] = 1500

# Cetak ulang seluruh dictionary
print("spesifikasi motor setelah perubahan RPM:")
print(motor_spec)
```

```python
# latihan 6

pembangkit = [{"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
              {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
              {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "operasi"},]

# (a) Cetak nama unit kedua
print(f"Nama unit kedua: {pembangkit[1]['nama']}")

# (b) ubah status salah satu (misal, Unit-1 menjadi 'non-operasi')
pembangkit[0]['status'] = "non-operasi"
print(f"Daftar unit-1 setelah di ubah: {pembangkit[0]['status']}")

# (c) Tambahkan satu unit baru ke list dengan append()
pembangkit.append({"nama": "unit-4", "kapasitas_MW": 120, "status": "uji coba"})

# (d) Cetak seluruh data
for unit in pembangkit:
    print(unit)
```
