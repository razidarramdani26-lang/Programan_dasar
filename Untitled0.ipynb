```python
suhu_saat_ini = 60

while suhu_saat_ini < 85:
    suhu_saat_ini += 5
    print(f"Suhu saat ini: {suhu_saat_ini} C")
    if suhu_saat_ini == 80:
        print("WARNING: Mendekati target!")

print("Target 85 C tercapai.")
```

```python
data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
tidak_valid = 0

for n, item in enumerate(data, start=1):
    try:
        nilai = float(item)
    except ValueError:
        print(f"Unit {n}: data tidak valid")
        tidak_valid += 1
        continue

    if nilai > 3.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")
        print("KRITIS: pemeriksaan dihentikan")
        break
    elif nilai > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")

print(f"Jumlah data tidak valid: {tidak_valid}")
```
